# Aanalysis: Who benefits from AI?

checking if AI adoption is associated with better outcomes using the following columns:

1. ai_adoption_level
2. productivity_score
3. salary_usd
4. annual_performance_rating
5. promotion_probability
5. job_satisfaction

```python
df["ai_adoption_level"].value_counts()
```

```python
df["ai_adoption_level"].unique()
df
```

```python
# comparing outcome by ai_aduption_level

df.groupby("ai_adoption_level")[[
    "productivity_score",
    "salary_usd",
    "promotion_probability",
    "job_satisfaction"
]].mean()
```

```python
# checking the missing values

df[[
    "ai_adoption_level",
    "productivity_score",
    "salary_usd",
    "promotion_probability",
    "job_satisfaction"
]].isna().sum()
```

```python
# average impact of ai on jobs.

ai_impact = df.groupby("ai_adoption_level")[[
    "productivity_score",
    "salary_usd",
    "promotion_probability",
    "job_satisfaction"
]].mean()

ai_impact
```

```python
# grouping impact to see status

df["ai_adoption_group"] = pd.cut(
    df["ai_adoption_level"],
    bins=[0, 25, 50, 75, 100],
    labels=["Low", "Medium", "High", "Very High"],
    include_lowest=True
)

# the actual grouping status
df["ai_adoption_group"].value_counts().sort_index()
```

```python
df["ai_adoption_group"].value_counts().sort_index()
```

```python
ai_impact = df.groupby("ai_adoption_group", observed=True)[[
    "productivity_score",
    "salary_usd",
    "promotion_probability",
    "job_satisfaction"
]].mean()

ai_impact
```

```python
print(ai_impact.to_string())
```

Key things to note here:

1. Productivity rises from Low to Very High: 42.24 to 56.04
That's about a 32.7% increase.

2. Salary rises dramatically from: $25,358 to $69,337
That's about a 173% increase.

3. Promotion probability rises from 36.9% to 45.2%
That's an increase of about 8.3 percentage points.

4. Job satisfaction also rises from 6.81 to 7.25
That is a smaller increase, but still upward.

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
ai_impact["productivity_score"].plot(kind="bar")

plt.title("AI Adoption vs Productivity")
plt.xlabel("AI Adoption Level")
plt.ylabel("Average Productivity Score")
plt.xticks(rotation=0)
plt.show()
```

```python
ai_impact["salary_usd"].plot(kind="bar")

plt.title("AI Adoption vs Salary")
plt.xlabel("AI Adoption Level")
plt.ylabel("Average Salary (USD)")
plt.xticks(rotation=0)
plt.show()
```

```python
ai_impact["promotion_probability"].plot(kind="bar")

plt.title("AI Adoption vs Promotion Probability")
plt.xlabel("AI Adoption Level")
plt.ylabel("Average Promotion Probability")
plt.xticks(rotation=0)
plt.show()
```

```python
ai_impact["job_satisfaction"].plot(kind="bar")

plt.title("AI Adoption vs Job Satisfaction")
plt.xlabel("AI Adoption Level")
plt.ylabel("Average Job Satisfaction")
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    x=ai_impact.index,
    y=ai_impact["salary_usd"]
)

plt.title("AI Adoption vs Salary", fontsize=16)
plt.xlabel("AI Adoption Level")
plt.ylabel("Average Salary (USD)")
plt.xticks(rotation=0)
plt.ticklabel_format(style="plain", axis="y")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.ticker as mtick

plt.figure(figsize=(9, 5))

sns.barplot(
    x=ai_impact.index,
    y=ai_impact["promotion_probability"]
)

plt.title("AI Adoption vs Promotion Probability", fontsize=16)
plt.xlabel("AI Adoption Level")
plt.ylabel("Promotion Probability")
plt.gca().yaxis.set_major_formatter(mtick.PercentFormatter(1))

plt.tight_layout()
plt.show()
```
