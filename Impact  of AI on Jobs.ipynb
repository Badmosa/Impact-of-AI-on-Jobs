# Analysis 2: Who is at risk from AI?

To identify who is at at ai risk, I identify the workers, industries, and job roles that are most exposed to AI-driven displacement using the following columns from the dataset:

1. automation_probability
2. layoff_probability
3. employment_status_after_ai
4. industry
5. job_role

```python
# using "groupby" to check average automation and layoff probabilities to check if worker risk increase or decrease as AI adoption increases

risk_by_ai = df.groupby("ai_adoption_group", observed=True)[[
    "automation_probability",
    "layoff_probability"
]].mean()

risk_by_ai
```

```python
# Rounding "risk_by_ai" output for better view and understanding.

risk_by_ai_percentage = (risk_by_ai * 100).round(1)

risk_by_ai_percentage
```

```python
df["employment_status_after_ai"].value_counts()
```

```python
df["employment_status_after_ai"].value_counts(normalize=True) * 100
```

```python
# Find the most vulnerable job roles


risk_by_role = df.groupby("job_role")[[
    "automation_probability",
    "layoff_probability"
]].mean().sort_values(
    "automation_probability",
    ascending=False
)

risk_by_role.head(5)
```

```python
risk_by_role_percentage = (risk_by_role * 100).round(1)

risk_by_role_percentage
```

```python
risk_by_role_percentage.head(10).style.format("{:.1f}%")
```

```python
# Final Top 5 highest ranking job roles

top_risk = risk_by_role.sort_values(
    "automation_probability",
    ascending=False
).head()

top_risk_percentage = (top_risk * 100).round(1)

top_risk_percentage
```

```python
# chart

plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_risk_percentage.reset_index(),
    x="automation_probability",
    y="job_role"
)

plt.title("Top 5 Jobs by AI Automation Risk")
plt.xlabel("Automation Probability by %")
plt.ylabel("Job Role")

plt.tight_layout()
plt.show()
```

```python
# Finding overall risk

import numpy as np

risk_by_role["overall_risk"] = np.mean(
    [
        risk_by_role["automation_probability"],
        risk_by_role["layoff_probability"]
    ],
    axis=0
)
```

```python
# Overall ai job risk

Overall_ai_job_risk = risk_by_role.sort_values(
    "overall_risk",
    ascending=False
).head(10)

print(Overall_ai_job_risk)
```

```python
# Overall risk formatting

Overall_ai_job_risk.style.format("{:.1f}%")
```

```python
# Rounding overall_ai_job_risk value.

Overall_ai_job_risk = (risk_by_role["overall_risk"] * 100).round(1)
```

```python
# chart for the jobs at risk frisk from AI

top_5 = Overall_ai_job_risk.sort_values(ascending=False).head()

plt.figure(figsize=(11, 7))

sns.barplot(
    x=top_5.values,
    y=top_5.index
)

plt.title("Top Jobs at Risk from AI")
plt.xlabel(" ")
plt.ylabel(" ")

plt.xlim(0, top_5.max() + 5)

for i, value in enumerate(top_5.values):
    plt.text(
        value + 0.5,
        i,
        f"{value:.1f}%",
        va="center"
    )

plt.tight_layout()
plt.show()
```

```python
# value check

df["reskilling_completed"].value_counts()
```

```python
# comparing the re-skilled and non-reskilled workers

reskill_impact = df.groupby("reskilling_completed")[[
    "training_hours",
    "automation_probability",
    "layoff_probability",
    "salary_usd",
    "promotion_probability"
]].mean()

reskill_impact
```

```python
# more comparism between skilled and non-reskilled workers
pd.crosstab(
    df["reskilling_completed"],
    df["employment_status_after_ai"],
    normalize="index"
) * 100
```

```python

reskill_outcomes = pd.crosstab(
    df["reskilling_completed"],
    df["employment_status_after_ai"],
    normalize="index"
) * 100

difference = np.subtract(
    reskill_outcomes.loc["Yes"],
    reskill_outcomes.loc["No"]
)

difference.round(1)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

employment_outcomes = pd.crosstab(
    df["reskilling_completed"],
    df["employment_status_after_ai"],
    normalize="index"
) * 100

employment_outcomes = employment_outcomes.reset_index()

employment_outcomes = employment_outcomes.melt(
    id_vars="reskilling_completed",
    var_name="employment_status",
    value_name="percentage"
)

plt.figure(figsize=(11, 6))

sns.barplot(
    data=employment_outcomes,
    x="employment_status",
    y="percentage",
    hue="reskilling_completed"
)

plt.title("Employment Outcomes: Reskilled vs Non-Reskilled Workers"
         )

plt.xlabel("Employment Status")
plt.ylabel("Percentage of Workers")
plt.ylim(0, 100)

plt.legend(title="Reskilling Completed")

plt.tight_layout()
plt.show()
```

The above chart shows that reskilled workers have a much higher employment rate and dramatically lower unemployment rate.

# Analysis Question 4: Do workers with higher AI literacy earn more than those with lower AI literacy?


```python
# Creating an ai literacy group

df["ai_literacy_group"] = pd.cut(
    df["ai_literacy_score"],
    bins=[0, 25, 50, 75, 100],
    labels=["Low", "Medium", "High", "Expert"],
    include_lowest=True
)

df
```

```python
# Comparing each of the liteacy gorups

salary_premium = df.groupby("ai_literacy_group", observed=True)[[
    "salary_usd",
    "promotion_probability",
    "digital_skill_score"
]].mean()

salary_premium
```

```python
# formating 
salary_premium.style.format({
    "salary_usd": "${:,.0f}",
    "promotion_probability": "{:.1%}",
    "digital_skill_score": "{:.1f}"
})
```

```python


low_salary = salary_premium.loc["Low", "salary_usd"]
expert_salary = salary_premium.loc["Expert", "salary_usd"]

salary_premium_pct = np.divide(
    expert_salary - low_salary,
    low_salary
) * 100

print(f"AI salary premium: {salary_premium_pct:.1f}%")
```

```python
# chart for ai literacy vs average salary

salary_premium = salary_premium.sort_values(
    "salary_usd",
    ascending=False
)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=salary_premium.reset_index(),
    x="ai_literacy_group",
    y="salary_usd",
    order=salary_premium.index
)

plt.title("AI Literacy vs Average Salary", fontsize=16)
plt.xlabel("AI Literacy Level")
plt.ylabel("Average Salary")

plt.ticklabel_format(style="plain", axis="y")

for i, value in enumerate(salary_premium["salary_usd"]):
    plt.text(
        i,
        value + 2000,
        f"${value:,.0f}",
        ha="center",
        fontweight="bold"
    )

plt.tight_layout()
plt.show()
```

# Analysis Question 5: What is the human cost?

We want to examine whether higher AI adoption is associated with higher stress and lower job satisfaction

```python
# Calculate stress and satisfaction by AI adoption group
human_cost = df.groupby("ai_adoption_group", observed=True)[[
    "mental_stress_score",
    "job_satisfaction"
]].mean()

human_cost
```

```python
# Round the results
human_cost = human_cost.round(1)

human_cost
```

```python
# what does it cost to higher humans?

human_cost["human_cost_index"] = (
    human_cost["mental_stress_score"] -
    human_cost["job_satisfaction"]
)

human_cost
```

```python
# chart for the level of workers "mental stress" compared with "ai adoption".

plt.figure(figsize=(10, 6))

sns.lineplot(
    data=human_cost.reset_index(),
    x="ai_adoption_group",
    y="mental_stress_score",
    marker="o",
    linewidth=2.5,
    label="Mental Stress"
)

plt.title(
    "Level of Mental Stress",
    fontsize=17,
    fontweight="bold"
)

plt.xlabel(" ")
plt.ylabel(" ")

plt.tight_layout()
plt.show()
```

```python
# chart for the level of workers "job certisfaction" compared with ai "adoption".

plt.figure(figsize=(10, 6))

sns.lineplot(
    data=human_cost.reset_index(),
    x="ai_adoption_group",
    y="job_satisfaction",
    marker="o",
    linewidth=2.5,
    label="Job Satisfaction"
)

plt.title(
    "Job Satisfaction",
    fontsize=17,
    fontweight="bold"
)

plt.xlabel(" ")
plt.ylabel(" ")

plt.tight_layout()
plt.show()
```

Overall, a higher AI adoption was associated with lower mental stress and higher job satisfaction among workers in the dataset. This suggests that AI adoption may improve rather than worsen employee experience, although the analysis shows association, not causation.
