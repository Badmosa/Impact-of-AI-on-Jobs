```python
df["ai_adoption_group"].value_counts().sort_index()
```

```python
ai_impact = df.groupby("ai_adoption_group", observed=True)[[
    "productivity_score",
    "salary_usd",
    "promotion_probability",
    "job_satisfaction"
]].mean()

ai_impact
```

```python
print(ai_impact.to_string())
```

Key things to note here:

1. Productivity rises from Low to Very High: 42.24 to 56.04
That's about a 32.7% increase.

2. Salary rises dramatically from: $25,358 to $69,337
That's about a 173% increase.

3. Promotion probability rises from 36.9% to 45.2%
That's an increase of about 8.3 percentage points.

4. Job satisfaction also rises from 6.81 to 7.25
That is a smaller increase, but still upward.

# Analysis 2: Who is at risk from AI?

To identify who is at at ai risk, I identify the workers, industries, and job roles that are most exposed to AI-driven displacement using the following columns from the dataset:

1. automation_probability
2. layoff_probability
3. employment_status_after_ai
4. industry
5. job_role

```python
# using "groupby" to check average automation and layoff probabilities to check if worker risk increase or decrease as AI adoption increases

risk_by_ai = df.groupby("ai_adoption_group", observed=True)[[
    "automation_probability",
    "layoff_probability"
]].mean()

risk_by_ai
```

```python
# Rounding "risk_by_ai" output for better view and understanding.

risk_by_ai_percentage = (risk_by_ai * 100).round(1)

risk_by_ai_percentage
```

```python
df["employment_status_after_ai"].value_counts()
```

```python
df["employment_status_after_ai"].value_counts(normalize=True) * 100
```

```python
# Find the most vulnerable job roles


risk_by_role = df.groupby("job_role")[[
    "automation_probability",
    "layoff_probability"
]].mean().sort_values(
    "automation_probability",
    ascending=False
)

risk_by_role.head(5)
```

```python
risk_by_role_percentage = (risk_by_role * 100).round(1)

risk_by_role_percentage
```

```python
risk_by_role_percentage.head(10).style.format("{:.1f}%")
```

```python
# Final Top 5 highest ranking job roles

top_risk = risk_by_role.sort_values(
    "automation_probability",
    ascending=False
).head()

top_risk_percentage = (top_risk * 100).round(1)

top_risk_percentage
```

```python
# chart

plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_risk_percentage.reset_index(),
    x="automation_probability",
    y="job_role"
)

plt.title("Top 5 Jobs by AI Automation Risk")
plt.xlabel("Automation Probability by %")
plt.ylabel("Job Role")

plt.tight_layout()
plt.show()
```

```python
# Finding overall risk

import numpy as np

risk_by_role["overall_risk"] = np.mean(
    [
        risk_by_role["automation_probability"],
        risk_by_role["layoff_probability"]
    ],
    axis=0
)
```

```python
# Overall risk

Overall_risk = risk_by_role.sort_values(
    "overall_risk",
    ascending=False
).head(10)

print(Overall_risk)
```

```python
# Overall risk formatting

Overall_risk.style.format("{:.1f}%")
```

```python
Overall_risk = (risk_by_role["overall_risk"] * 100).round(1)
```

```python
top_5 = Overall_risk.sort_values(ascending=False).head()

plt.figure(figsize=(11, 7))

sns.barplot(
    x=top_5.values,
    y=top_5.index
)

plt.title("Top Jobs at Risk from AI")
plt.xlabel(" ")
plt.ylabel(" ")

plt.xlim(0, top_5.max() + 5)

for i, value in enumerate(top_5.values):
    plt.text(
        value + 0.5,
        i,
        f"{value:.1f}%",
        va="center"
    )

plt.tight_layout()
plt.show()
```
